# SCALING

```python
import pandas as pd
df = pd.read_csv('lalaiga 2 all player final - lalaiga 2 all player.csv')
```

```python
df.info()
```

```python
# birinchi missing valuelarni to'ldirib olamiz
```

```python
for col in df.columns:
    if df[col].isnull().sum() > 0:
        if df[col].dtype != 'str':
           df[col] = df[col].fillna(df[col].mean())
        else:
            df[col] = df[col].fillna(df[col].mode()[0])
```

```python
df.info()
```

# ENCODING

```python
from sklearn.preprocessing import LabelEncoder
encoder = LabelEncoder()
encoder
for col in df.columns:
    if df[col].dtype == 'str':
        if df[col].nunique() >= 5:
            df[col] = encoder.fit_transform(df[col])
        else:
            dummies = pd.get_dummies(df[col],prefix='col',dtype=int)
            df = pd.concat(df[col].drop(columns = col),dummies,axis=1)
```

```python
df.info()
```

# SCALING

```python
df.head()
```

```python
from sklearn.preprocessing import StandardScaler,MinMaxScaler,RobustScaler
```

```python
num_col = df.select_dtypes(include=['int64','float64'])
num_col
```

# STANDART SCALER

```python
scaler = StandardScaler()
scaler
```

```python
df['Player Name'] = scaler.fit_transform(df[['Player Name']])
```

```python
df.head()
```

```python
#MIN MAX SCALER
```

```python
scaler = MinMaxScaler()
scaler
```

```python
df['YC'] = scaler.fit_transform(df[['YC']])
```

```python
df.head()
```

# ROBUSTSCALER

```python
scaler = RobustScaler()
scaler
```

```python
df['MinP'] = scaler.fit_transform(df[['MinP']])
```

```python
df.head()
```

# For Loop + Scaling

```python
scaler = MinMaxScaler()
for col in df.columns:
    if df[col].dtype != 'object' and col != 'Rating':
        df[col] = scaler.fit_transform(df[[col]])
```

```python
df.head()
```

```python
# RATING bizda Target bo'lgani uchun uni scaling qilmaymiz
scaler = RobustScaler()
for col in num_col:
    df[col] = scaler.fit_transform(df[[col]])
```
