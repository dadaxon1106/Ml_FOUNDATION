# Data Preprocessing + for loop

```python
# MISSIN + FOR LOOP
```

```python
df.isnull().sum()
```

```python
# AGE yaroqsiz bo'lgani uchun drop qilamiz
df.drop('Age',axis=1,inplace=True)
```

```python
df.info()
```

```python
# MODE ORQALI
df.fillna({'AI_Tool':df['AI_Tool'].mode()[0]},inplace=True)
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
# MEAN 
df.fillna({'Productivity_Score':df['Productivity_Score'].mean()},inplace=True)
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
df.fillna({'Satisfaction_Score':df['Satisfaction_Score'].median()},inplace=True)
```

```python
df.isnull().sum()
```

```python
# FIXED USULI bilan
df.fillna({'AI_Purpose':'Coding'},inplace=True)
```

```python
df.isnull().sum()
```

```python
# LOOP ORQALI TO'LDIRISH
```

```python
# COLUMNLARINI TOPISH
df.columns
```

```python
for col in df.columns:
    if df[col].isnull().any():
        if df[col].dtype == 'str':
            df[col] = df[col].fillna(df[col].mode()[0])
        else:
            df[col] = df[col].fillna(df[col].mean())
```

```python
df.isnull().sum()
```

```python

# MEDIAN VA FIXED ORQ;ALI
for col in df.columns:
        if df[col].isnull().sum() != 0:
            if df[col].dtype == 'float64':
                 df[col] = df[col].fillna(df[col].median())
            else:
                df[col] = df[col].fillna("Coding")
```

```python
df.info()
```

```python
# DROP ORQALI
for col in df.columns:
    if df[col].isnull().sum() / len(df) > 0.5:
        df = df.drop(columns = [col])
```

```python
for col in df.columns:
    if df[col].dtype == 'str':
        if df[col].nunique() <= 5:
            dummies = pd.get_dummies(df[col],prefix='col',dtype=int)
            df = pd.concat([df.drop(columns=col),dummies],axis=1)
        else:
            df[col] = encoder.fit_transform(df[col])
```
