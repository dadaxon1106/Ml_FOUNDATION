```python
df.drop('Id',axis = 1,inplace = True) 
df.drop('Year_Birth',axis = 1,inplace = True) 
# ID va Year_Birth ustuni predict xususiyati bo'lmagani uchun drop qili yuboramiz
```

```python
df.info()
```

```python
# MISSING THE VALUES BY FOR LOOP
# Tushirib qoldirilgan qiymatlarni Loop orqali to'ldirish
for col in df.columns:
    if df[col].isnull().any():
        if df[col].dtype == 'str':
            df[col] = df[col].fillna(df[col].mode()[0])
        else:
            df[col] = df[col].fillna(df[col].mean())
```

```python
df.info()
```

# ENCODING

```python
from sklearn.preprocessing import LabelEncoder
encoder = LabelEncoder()
# encoder o'zgaruvchisiga LabelEncoderni call qilish
```

```python
df.nunique()
```

```python
# Loop orqali encoding qilish jarayoni
for col in df.columns:
    if df[col].dtype == 'str':
        if df[col].nunique() >= 5:
            df[col] = encoder.fit_transform(df[col])
        else:
            dummies = pd.get_dummies(df[col],prefix = 'col',dtype = int)
            df = pd.concat(df[col].drop(columns = col),dummies,axis = 1)
```

```python
df.info()
```

```python
df.drop('Education',axis = 1,inplace=True)
```

# SCALING

```python
num_col = df.columns.drop('Response') ## Predict qilinadigan column scaling qilinmaydi
```

```python
num_col
```

```python
from sklearn.preprocessing import MinMaxScaler # SCALING QILISH UCHUN MINMAXSCALER KUTUBXONASINI ORNATISH
scaler = MinMaxScaler()
```

```python
# For loop orqali scaling qilish
for col in num_col:
    df[col] = scaler.fit_transform(df[[col]])
```
