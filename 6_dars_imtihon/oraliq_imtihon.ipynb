```python
import pandas as pd
df = pd.read_csv("Titanic-Dataset.csv")
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
# MODE ORQALI To'LDIRISH
df.fillna({'Age':df['Age'].mean()},inplace=True)
```

```python
df.isnull().sum()
```

```python
# Mode orqali to'ldirish
df.fillna({'Embarked':df['Embarked'].mode()[0]},inplace=True)
```

```python
df.isnull().sum()
```

```python
# FIXED orqali to'ldirish
df.fillna({'Cabin':'D32'},inplace=True)
```

```python
df.isnull().sum()
```

```python
df.info()
```
