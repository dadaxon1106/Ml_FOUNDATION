# Tushirib qoldirilgan qiymatlarni to'ldirish

```python
import pandas as pd
df = pd.read_csv("bundesliga_player.csv")
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
# MEAN ORQALI TUSHIRIB QOLDIRILGAN QIYMATLARNI TO'LDIRISH
df.fillna({'price': df['price'].mean()}, inplace=True)
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
# MEDIAN USULI ORQALI TO'LDIRISH
df.fillna({'max_price': df['max_price'].median()}, inplace=True)
```

```python
df.info()
```

```python
# MODE ORQALI 
df.fillna({'outfitter':df['outfitter'].mode()[0]}, inplace=True)
```

```python
df.info()
```

```python
# FIXED USULI ORQALI
df.fillna({'foot': 'LEFT'}, inplace=True)
```

```python
df.info()
```

```python
# DROP USULI ORQALI
df.drop('player_agent',axis=1,inplace=True)
```

```python
df.info()
```
