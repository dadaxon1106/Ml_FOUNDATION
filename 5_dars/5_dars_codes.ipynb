# ENCODING

```python
import pandas as pd
df = pd.read_csv("wcplayerstatistics2014.csv")
```

```python
df.info()
```

# One - Hot Encoding

```python
# One - Hot encoding bu categorical qiymatdan numerical qiymatga o'tkazishdir
# Bu jarayonda str(objectda klasslar soniga qarab ustun yaratadi va 0 va 1 ga almashtirib chiqadi)
# bunda avval dastlab classlar sonini aniqlaymiz keyin dummylarni yaratishdan boshlaymiz
df.nunique()
```

```python
# demak biz ko'rib oldik qancha classlar borligini 
dummies = pd.get_dummies(df['Pos'],prefix='col',dtype=int)
```

```python
dummies
```

```python
df.info()
```

```python
# endi yangi yaratilgan yangi qiymatni eskisiga qo'shamiz
df = pd.concat([df.drop(columns=['Pos']),dummies],axis=1)
```
