```python

products = {}
is_continue = True
all_sum = 0
count = 0
while is_continue:
  product = input("Mahsulot nomini kiriting: ")
  price_product = int(input("Narxi: "))
  products[product] = price_product
  count +=1
  if count == 3:
    is_continue = False
for i in products.values():
  if i > 500000:
    all_sum += i - 1000000
  elif i > 200000:
    all_sum += i - 20000
  else:
    all_sum += i

print(f"Umumiy narx (%) bilan: {all_sum}")
```

```python
from typing import Tuple
# MEDIAN ORQALI TO'lDIRISH
df['Calories_Consumed'].fillna(df['Calories_Consumed'].median(),inplace=True)
```

```python
df.isnull().sum()
```

```python
# MODE orqali to'ldirish
df['Workout_Type'].fillna(df['Workout_Type'].mode()[0],inplace=True)
```

```python
# MEAN orqali to'ldirish
df['Calories_Consumed'].fillna(df['Calories_Consumed'].mean(),inplace=True)
```
