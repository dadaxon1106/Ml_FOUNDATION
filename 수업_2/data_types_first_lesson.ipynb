Amallar (Operations)

```python
# Arifmetik amallar: qo'shish, ayirish, ko'paytirish va darajaga oshirish
x = 6
y = 2

print(x + y)  # Qo'shish
print(x - y)  # Ayirish
print(x * y)  # Ko'paytirish
print(x ** y) # Darajaga oshirish
```

```python
# Qoldiq (%), oddiy bo'lish (/) va butun bo'lish (//)
x = 17
y = 5
print(x % y)   # Bo'lishdan qolgan qoldiq
print(x / y)   # Oddiy bo'lish
print(x // y)  # Butun bo'lish
```

Mantiqiy operatorlar (Logic Operators)

```python
# and — ikkala shart ham rost bo'lsa True
# or — shartlardan kamida bittasi rost bo'lsa True
rang = 'ko\'k'
print(2 < 5 and 5 < 9)
print(rang == 'ko\'k' or rang == 'qizil')
```

Mustaqil mashqlar

```python
# 1-MASHQ: Matnning birinchi va oxirgi harfini chiqarish
soz = 'maktab'
birinchi_harf = soz[0]
oxirgi_harf = soz[-1]
print(f'Birinchi harf: {birinchi_harf}')
print(f'Oxirgi harf: {oxirgi_harf}')
```

```python
# 2-MASHQ: Son musbatmi va juftmi?
son = 14
print(son > 0)       # Musbatligini tekshirish
print(son % 2 == 0)  # Juftligini tekshirish
```

```python
# 3-MASHQ: Matnning oxiridan ikkinchi belgisini chiqarish
shahar = 'Toshkent'
oxiridan_ikkinchi = shahar[-2]
print(oxiridan_ikkinchi)

# 4-MASHQ: Son musbatmi va toqmi?
son = -35
print(son > 0)       # Musbatligini tekshirish
print(son % 2 != 0)  # Toqligini tekshirish
```
